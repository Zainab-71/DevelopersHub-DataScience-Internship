# Task 7: Customer Segmentation Using Unsupervised Learning

## Introduction

This notebook is part of my Data Science & Analytics Internship at
DevelopersHub Corporation. The objective of this task is to build an
unsupervised machine learning model that segments mall customers into
distinct groups based on their spending habits and personal attributes,
and to propose targeted marketing strategies for each identified segment.

## Problem Statement

I am working with the Mall Customers Dataset. It contains 2,240 customer
records with 29 columns including annual income, spending across product
categories, marital status, education level, and number of children.
The target is not a label but a pattern — the goal is to clean the data,
explore spending behaviour through visualization, apply K-Means Clustering
to discover natural customer groupings, reduce dimensions using PCA and
t-SNE for visual confirmation, and recommend data-driven marketing
strategies for each segment.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, silhouette_samples
```

```python
# Plot styling
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['font.size'] = 12
sns.set_style('whitegrid')
sns.set_palette('Set2')

print("All libraries loaded successfully.")
```

```python
# Load the dataset (tab-separated)
df = pd.read_csv('marketing_campaign.csv', sep='\t')

print(f"Dataset Shape: {df.shape}")
print(f"Feature Overview:")
print(df.dtypes)
df.head()
```

```python
df.describe
```

```python

df.drop(columns=['ID', 'Z_CostContact', 'Z_Revenue'], inplace=True)

# Handle missing values
print(f"Missing values before: {df['Income'].isnull().sum()}")
df['Income'].fillna(df['Income'].median(), inplace=True)
print(f"Missing values after:  {df['Income'].isnull().sum()}")
```

```python
# Feature Engineering 
# Age from birth year
df['Age'] = 2024 - df['Year_Birth']

# Total amount spent across all product categories
df['TotalSpending'] = (df['MntWines'] + df['MntFruits'] + df['MntMeatProducts'] +
                       df['MntFishProducts'] + df['MntSweetProducts'] + df['MntGoldProds'])

# Total number of purchases across all channels
df['TotalPurchases'] = (df['NumWebPurchases'] + df['NumCatalogPurchases'] +
                        df['NumStorePurchases'] + df['NumDealsPurchases'])

# Family size (partner + children)
df['Family_Size'] = (df['Marital_Status'].isin(['Married', 'Together']).astype(int) +
                     df['Kidhome'] + df['Teenhome'])

# Total campaigns accepted
df['TotalCampaignsAccepted'] = (df['AcceptedCmp1'] + df['AcceptedCmp2'] +
                                 df['AcceptedCmp3'] + df['AcceptedCmp4'] +
                                 df['AcceptedCmp5'] + df['Response'])

# Customer tenure in days
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], dayfirst=True)
df['Customer_Days'] = (pd.Timestamp('2024-01-01') - df['Dt_Customer']).dt.days
```

```python

# Encode categoricals
# Simplify education
edu_map = {'Basic': 0, '2n Cycle': 1, 'Graduation': 2, 'Master': 3, 'PhD': 4}
df['Education_Enc'] = df['Education'].map(edu_map)

# Simplify marital status
df['Is_Partner'] = df['Marital_Status'].isin(['Married', 'Together']).astype(int)

# Remove outliers
before = len(df)
df = df[df['Income'] < 600000]
df = df[df['Age'] < 100]
print(f"\nRows removed as outliers: {before - len(df)}")
print(f"Clean dataset shape: {df.shape}")
```

```python
# Exploratory Data Analysis
# Income Distribution
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].hist(df['Income'], bins=40, color='steelblue', edgecolor='white')
axes[0].set_title('Income Distribution')
axes[0].set_xlabel('Annual Income ($)')
axes[0].set_ylabel('Count')

axes[1].hist(df['TotalSpending'], bins=40, color='coral', edgecolor='white')
axes[1].set_title('Total Spending Distribution')
axes[1].set_xlabel('Total Spending ($)')

axes[2].hist(df['Age'], bins=30, color='mediumseagreen', edgecolor='white')
axes[2].set_title('Age Distribution')
axes[2].set_xlabel('Age (years)')

plt.tight_layout()
plt.savefig('eda_distributions.png', dpi=150, bbox_inches='tight')
plt.show()
print("Key stats:")
print(df[['Income', 'TotalSpending', 'Age']].describe().round(2))
```

```python
# Spending breakdown by category
spend_cols = ['MntWines', 'MntFruits', 'MntMeatProducts',
              'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']
spend_totals = df[spend_cols].sum().sort_values(ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

spend_totals.plot(kind='bar', ax=axes[0], color=sns.color_palette('Set2', 6), edgecolor='white')
axes[0].set_title('Total Spending by Product Category')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Total Spent ($)')
axes[0].tick_params(axis='x', rotation=30)

# Income vs Spending scatter
axes[1].scatter(df['Income'], df['TotalSpending'], alpha=0.4, color='steelblue', s=15)
axes[1].set_title('Income vs Total Spending')
axes[1].set_xlabel('Annual Income ($)')
axes[1].set_ylabel('Total Spending ($)')

plt.tight_layout()
plt.savefig('eda_spending.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Correlation heatmap
corr_cols = ['Income', 'Age', 'TotalSpending', 'TotalPurchases',
             'Family_Size', 'TotalCampaignsAccepted', 'Recency', 'Customer_Days']

plt.figure(figsize=(10, 8))
corr = df[corr_cols].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.5)
plt.title('Feature Correlation Heatmap', fontsize=14, pad=12)
plt.tight_layout()
plt.savefig('eda_correlation.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Education & Marital Status breakdown 
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

df['Education'].value_counts().plot(kind='bar', ax=axes[0],
    color=sns.color_palette('pastel', 5), edgecolor='white')
axes[0].set_title('Education Level Distribution')
axes[0].tick_params(axis='x', rotation=30)

df['Marital_Status'].value_counts().plot(kind='pie', ax=axes[1],
    autopct='%1.1f%%', startangle=90,
    colors=sns.color_palette('Set3', df['Marital_Status'].nunique()))
axes[1].set_title('Marital Status Distribution')
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('eda_demographics.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Feature Selection & Scaling
# Select features most relevant for customer segmentation
features = [
    'Income', 'Age', 'TotalSpending', 'TotalPurchases',
    'Family_Size', 'TotalCampaignsAccepted', 'Recency',
    'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases',
    'Education_Enc', 'Customer_Days'
]

X = df[features].copy()
print("Feature set shape:", X.shape)
print("\nFeature means before scaling:")
print(X.mean().round(2))

# Standardize (KMeans is distance-based — scaling is mandatory)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=features)

print("\nFeature means after scaling (should all be ~0):")
print(X_scaled_df.mean().round(4))
```

```python
# Finding Optimal Number of Clusters
# Elbow Method
inertias = []
K_range = range(2, 11)

for k in K_range:
    km = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    km.fit(X_scaled)
    inertias.append(km.inertia_)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

axes[0].plot(K_range, inertias, 'bo-', markersize=8, linewidth=2)
axes[0].set_xlabel('Number of Clusters (K)')
axes[0].set_ylabel('Inertia (WCSS)')
axes[0].set_title('Elbow Method — Optimal K')
axes[0].axvline(x=4, color='red', linestyle='--', label='K=4 (elbow)')
axes[0].legend()
```

```python
# Silhouette Scores
sil_scores = []
for k in K_range:
    km = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    labels = km.fit_predict(X_scaled)
    sil_scores.append(silhouette_score(X_scaled, labels))

axes[1].plot(K_range, sil_scores, 'rs-', markersize=8, linewidth=2)
axes[1].set_xlabel('Number of Clusters (K)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Silhouette Score — Optimal K')
best_k = K_range[np.argmax(sil_scores)]
axes[1].axvline(x=best_k, color='blue', linestyle='--', label=f'Best K={best_k}')
axes[1].legend()

plt.tight_layout()
plt.savefig('optimal_k.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"Best K by Silhouette Score: {best_k} (score: {max(sil_scores):.4f})")
```

```python
# K-Means Clustering
# Train final KMeans with K=4
OPTIMAL_K = 4

kmeans = KMeans(n_clusters=OPTIMAL_K, init='k-means++', n_init=20, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print(f"K-Means trained with K={OPTIMAL_K}")
print(f"Final Inertia: {kmeans.inertia_:.2f}")
print(f"Silhouette Score: {silhouette_score(X_scaled, df['Cluster']):.4f}")
print(f"\nCluster Distribution:")
print(df['Cluster'].value_counts().sort_index())
```

```python
# Silhouette Plot
fig, ax = plt.subplots(figsize=(10, 6))

y_lower = 10
sil_vals = silhouette_samples(X_scaled, df['Cluster'])
colors = cm.Set2(np.linspace(0, 0.8, OPTIMAL_K))

for i in range(OPTIMAL_K):
    cluster_sil = np.sort(sil_vals[df['Cluster'] == i])
    size = cluster_sil.shape[0]
    y_upper = y_lower + size
    ax.fill_betweenx(np.arange(y_lower, y_upper), 0, cluster_sil,
                     facecolor=colors[i], edgecolor='none', alpha=0.8)
    ax.text(-0.05, y_lower + 0.5 * size, f'C{i}', fontsize=11, fontweight='bold')
    y_lower = y_upper + 10

avg_sil = silhouette_score(X_scaled, df['Cluster'])
ax.axvline(x=avg_sil, color='red', linestyle='--', label=f'Avg score: {avg_sil:.3f}')
ax.set_xlabel('Silhouette Coefficient')
ax.set_ylabel('Cluster')
ax.set_title('Silhouette Plot for K=4 Clusters')
ax.legend()
plt.tight_layout()
plt.savefig('silhouette_plot.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Dimensionality Reduction — PCA
# PCA to 2D for visualization
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

df['PCA1'] = X_pca[:, 0]
df['PCA2'] = X_pca[:, 1]

explained = pca.explained_variance_ratio_
print(f"PCA Component 1 explains: {explained[0]*100:.1f}% variance")
print(f"PCA Component 2 explains: {explained[1]*100:.1f}% variance")
print(f"Total variance explained: {sum(explained)*100:.1f}%")
```

```python
# Scree plot
pca_full = PCA(random_state=42).fit(X_scaled)
cumvar = np.cumsum(pca_full.explained_variance_ratio_)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))
# PCA cluster scatter
palette = sns.color_palette('Set2', OPTIMAL_K)
for c in range(OPTIMAL_K):
    mask = df['Cluster'] == c
    axes[0].scatter(df.loc[mask, 'PCA1'], df.loc[mask, 'PCA2'],
                    s=20, alpha=0.6, color=palette[c], label=f'Cluster {c}')
axes[0].set_title(f'PCA Visualization — K-Means Clusters\n(Explains {sum(explained)*100:.1f}% variance)')
axes[0].set_xlabel(f'PC1 ({explained[0]*100:.1f}%)')
axes[0].set_ylabel(f'PC2 ({explained[1]*100:.1f}%)')
axes[0].legend(title='Cluster')

# Scree plot
axes[1].plot(range(1, len(cumvar)+1), cumvar, 'bo-', linewidth=2, markersize=6)
axes[1].axhline(y=0.80, color='red', linestyle='--', label='80% variance')
axes[1].axhline(y=0.90, color='green', linestyle='--', label='90% variance')
axes[1].set_xlabel('Number of Components')
axes[1].set_ylabel('Cumulative Explained Variance')
axes[1].set_title('PCA Scree Plot')
axes[1].legend()

plt.tight_layout()
plt.savefig('pca_visualization.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Dimensionality Reduction — t-SNE
# t-SNE for non-linear cluster visualization
print("Running t-SNE (this may take ~30 seconds)...")
tsne = TSNE(n_components=2, perplexity=40, max_iter=1000, random_state=42)
X_tsne = tsne.fit_transform(X_scaled)

df['TSNE1'] = X_tsne[:, 0]
df['TSNE2'] = X_tsne[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# t-SNE by cluster
for c in range(OPTIMAL_K):
    mask = df['Cluster'] == c
    axes[0].scatter(df.loc[mask, 'TSNE1'], df.loc[mask, 'TSNE2'],
                    s=25, alpha=0.7, color=palette[c], label=f'Cluster {c}')
axes[0].set_title('t-SNE Visualization — K-Means Clusters')
axes[0].set_xlabel('t-SNE Dimension 1')
axes[0].set_ylabel('t-SNE Dimension 2')
axes[0].legend(title='Cluster')

# t-SNE colored by Income
sc = axes[1].scatter(df['TSNE1'], df['TSNE2'], c=df['Income'],
                     cmap='YlOrRd', s=20, alpha=0.7)
plt.colorbar(sc, ax=axes[1], label='Annual Income ($)')
axes[1].set_title('t-SNE — Colored by Income')
axes[1].set_xlabel('t-SNE Dimension 1')
axes[1].set_ylabel('t-SNE Dimension 2')

plt.tight_layout()
plt.savefig('tsne_visualization.png', dpi=150, bbox_inches='tight')
plt.show()
print("t-SNE complete.")
```

```python
# Cluster Profiling & Analysis
# Compute mean profile per cluster
profile_cols = ['Income', 'Age', 'TotalSpending', 'TotalPurchases',
                'Family_Size', 'TotalCampaignsAccepted', 'Recency',
                'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases',
                'MntWines', 'MntMeatProducts', 'MntGoldProds']

cluster_profile = df.groupby('Cluster')[profile_cols].mean().round(1)
print("Cluster Profiles (Mean Values):")
cluster_profile

# Radar / Spider chart per cluster 
from matplotlib.patches import FancyArrowPatch
import matplotlib.patches as mpatches

radar_features = ['Income', 'TotalSpending', 'TotalPurchases',
                  'TotalCampaignsAccepted', 'Family_Size', 'Recency']

# Normalize to 0-1 for radar
radar_data = cluster_profile[radar_features].copy()
radar_norm = (radar_data - radar_data.min()) / (radar_data.max() - radar_data.min())

N = len(radar_features)
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
angles += angles[:1]

fig, axes = plt.subplots(1, OPTIMAL_K, figsize=(20, 5), subplot_kw=dict(polar=True))

for c in range(OPTIMAL_K):
    values = radar_norm.loc[c].tolist() + radar_norm.loc[c].tolist()[:1]
    axes[c].plot(angles, values, 'o-', linewidth=2, color=palette[c])
    axes[c].fill(angles, values, alpha=0.25, color=palette[c])
    axes[c].set_xticks(angles[:-1])
    axes[c].set_xticklabels(radar_features, size=9)
    axes[c].set_ylim(0, 1)
    axes[c].set_title(f'Cluster {c}', fontsize=13, fontweight='bold', pad=15)

plt.suptitle('Cluster Radar Profiles (Normalized)', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('cluster_radar.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Box plots: Income & Spending by Cluster
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, col, title in zip(axes,
    ['Income', 'TotalSpending', 'TotalPurchases'],
    ['Income by Cluster', 'Total Spending by Cluster', 'Total Purchases by Cluster']):
    df.boxplot(column=col, by='Cluster', ax=ax,
               patch_artist=True,
               boxprops=dict(facecolor='lightblue'),
               medianprops=dict(color='red', linewidth=2))
    ax.set_title(title)
    ax.set_xlabel('Cluster')
    plt.sca(ax)
    plt.title(title)

plt.suptitle('')
plt.tight_layout()
plt.savefig('boxplots.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Spending category breakdown per cluster
spend_cols = ['MntWines', 'MntFruits', 'MntMeatProducts',
              'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']

spend_by_cluster = df.groupby('Cluster')[spend_cols].mean()
spend_by_cluster.columns = ['Wines', 'Fruits', 'Meat', 'Fish', 'Sweets', 'Gold']

spend_by_cluster.plot(kind='bar', figsize=(14, 6),
                      colormap='Set2', edgecolor='white', width=0.7)
plt.title('Average Spending per Category by Cluster', fontsize=14)
plt.xlabel('Cluster')
plt.ylabel('Average Spend ($)')
plt.legend(title='Category', bbox_to_anchor=(1.01, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('spending_by_cluster.png', dpi=150, bbox_inches='tight')
plt.show()
```

## Conclusion and Key Insights

The correlation heatmap revealed that Income has the strongest relationship
with Total Spending at 0.79, making it by far the most influential factor
in separating customer groups. Every other feature scored below 0.35,
confirming that how much a customer earns is the primary driver of how
much they spend.

Visualizations confirmed this consistently:

1. The Income Distribution chart showed a roughly normal distribution
   centered around USD 52K, with a long right tail reaching USD 160K, indicating
   a small but high-value group of premium spenders.

2. The Total Spending Distribution was heavily right-skewed. The majority
   of customers spent under USD 500, but a small group spent up to USD 2,500.
   This confirmed the decision to use K-Means with standardized features
   rather than raw values.

3. The Spending by Category chart showed that Wines dominated all product
   categories by a significant margin, followed by Meat Products. Both
   categories were heavily concentrated in the high-income clusters.

4. The Education chart showed that Graduation-level customers made up the
   largest group, while PhD and Master holders showed higher average
   spending than Basic or 2n Cycle educated customers.

K-Means with K=4 was selected based on the Elbow Method, achieving a
Silhouette Score of 0.1294. Four distinct customer segments were
identified: High-Value Premium Shoppers (C0), Mid-Tier Active Buyers (C2),
Budget-Conscious Families (C1), and Low-Income Minimal Spenders (C3).
PCA and t-SNE both visually confirmed the separation between these groups.

**Final Takeaway:** Income and spending behaviour together define customer
value far more than any demographic feature. Cluster 0, despite being the
smallest group, carries the highest revenue potential per customer and
should be the primary focus for premium marketing campaigns. A more
balanced dataset with additional behavioural features such as browsing
history or loyalty points would further improve segmentation quality.
